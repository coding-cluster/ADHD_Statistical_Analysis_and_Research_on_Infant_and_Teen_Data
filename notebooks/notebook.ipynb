# 1. Introduction

Attention Deficit and Hyperactivity Disorder (ADHD) is classified as a neurodevelopment disorder that implicates problems to attention-span, controlling impulsive behaviour, and remaining in a state of little-to-no activity. (National Library of Medicine, n.d.)

This is one of the most common disorders affecting children, and it can lead to various problems such as low self-esteem and difficulties with their abilities when not properly treated. Failing to address it can have repercussions throughout life, where in adulthood, people with ADHD may experience low self-esteem, sensitivity to criticism, and increased self-criticism. (American Psychiatric Association, n.d.)

In the present work, an analysis design with a sampling plan will be defined to carry out a statistical analysis on datasets (without labeling or names of the participants) of children and adolescents with and without this disorder, in order to subsequently carry out a data statistical analysis process that allows us to understand this neurobiological condition.
